```python
import pandas as pd
import numpy as np
df = pd.read_csv("../data/raw/loan_book.csv")
```

```python
df['application_date'] = pd.to_datetime(
    df['application_date'],
    format='mixed'
)
```

```python
categorical_cols = [
    'home_ownership',
    'loan_purpose',
    'region',
    'email_domain_type',
    'phone_verified'
]
```

```python
for col in categorical_cols:
    df[col] = (
        df[col]
        .astype(str)
        .str.strip()
        .str.lower()
    )
```

```python
for col in categorical_cols:
    print(f"\n{col}")
    print(df[col].value_counts())
```

```python
df['missing_income'] = df['annual_income'].isnull().astype(int)

df['missing_employment_length'] = (
    df['employment_length_years']
    .isnull()
    .astype(int)
)

df['has_delinquency_history'] = (
    df['months_since_last_delinquency']
    .notnull()
    .astype(int)
)
```

```python
median_cols = [
    'annual_income',
    'employment_length_years',
    'num_open_accounts',
    'months_since_last_delinquency'
]
```

```python
for col in median_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
df['log_annual_income'] = np.log1p(df['annual_income'])

df['log_total_revolving_balance'] = np.log1p(
    df['total_revolving_balance']
)

df['log_loan_amount'] = np.log1p(df['loan_amount'])
```

```python
df['application_year'] = df['application_date'].dt.year

df['application_month'] = df['application_date'].dt.month

df['application_dayofweek'] = (
    df['application_date']
    .dt.day_name()
)
```

```python
drop_cols = [
    'applicant_id_hash',
    'application_date'
]
```

```python
df = df.drop(columns=drop_cols)
```

```python
df.to_csv(
    "../data/processed/feature_engineered_data.csv",
    index=False
)
```
